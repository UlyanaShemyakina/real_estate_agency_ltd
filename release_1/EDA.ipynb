```python
df_rub = df_raw[df_raw["Валюта"] == "руб."].copy()

print("Всего объявлений:", len(df_raw))
print("Объявлений в рублях:", len(df_rub))
print("Исключено объявлений в других валютах:", len(df_raw) - len(df_rub))
```

```python
df_rub["Цена"] = (
    df_rub["Цена"]
    .str.extract(r"^([\d.]+)")[0]
    .astype(float)
)

df_rub["Цена"].describe()
```

```python
df_rub["Общая площадь"] = (
    df_rub["Площадь, м2"]
    .str.split("/")
    .str[0]
    .astype(float)
)

df_rub["Цена за м2"] = (
    df_rub["Цена"] / df_rub["Общая площадь"]
)

df_rub["Цена за м2"].describe()
```
