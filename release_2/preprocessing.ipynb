```python
df["Валюта"] = df["Цена"].str.extract(r"^[\d.]+\s*([^\s/]+)")[0]
```

```python
df["Цена числом"] = df["Цена"].str.extract(r"^([\d.]+)")[0].astype(float)
```

```python
display(df[["Цена", "Валюта", "Цена числом"]].head(10))
```

```python
print("Пропуски в цене:", df["Цена числом"].isna().sum())
print("Пропуски в валюте:", df["Валюта"].isna().sum())
```

```python
df_rub = df[df["Валюта"] == "руб."].copy()
```

```python
print("Количество рублёвых объявлений:", len(df_rub))
```

```python
display(df_rub[["Цена", "Валюта", "Цена числом"]].head())
```

```python
parts_count = df_rub["Площадь, м2"].str.split("/").str.len()
display(parts_count.value_counts().sort_index())
```

```python
display(
    df_rub.loc[
        df_rub["Площадь, м2"].str.split("/").str.len() == 1,
        ["Площадь, м2", "Адрес", "Цена"]
    ].head(15)
)
```

```python
df_rub["Общая площадь"] = (
    df_rub["Площадь, м2"]
    .str.split("/")
    .str[0]
    .astype(float)
)
display(df_rub[["Площадь, м2", "Общая площадь"]].head(10))
```

```python
print("Пропуски в общей площади:", df_rub["Общая площадь"].isna().sum())
```
