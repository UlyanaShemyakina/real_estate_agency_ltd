```python
df_rub["rooms"] = (
    df_rub["Количество комнат"]
    .str.extract(r"^(\d+)")[0]
    .astype(float)
)

df_rub["room_layout"] = (
    df_rub["Количество комнат"]
    .str.extract(r",\s*(.+)$")[0]
)
```

```python
for column in df_rub.select_dtypes(include="number").columns:
    df_rub[column] = df_rub[column].fillna(df_rub[column].median())
```

```python
for column in df_rub.select_dtypes(include=["object", "str"]).columns:
    df_rub[column] = df_rub[column].fillna("unknown")
```

```python
print("Всего пропусков:", df_rub.isna().sum().sum())
```

```python
df_rub = df_rub.rename(columns={
    "ID  объявления": "id",
    "Количество комнат": "rooms_original",
    "Тип": "property_type",
    "Метро": "metro",
    "Адрес": "address",
    "Площадь, м2": "area_original",
    "Дом": "building",
    "Парковка": "parking",
    "Цена": "price_original",
    "Телефоны": "phones",
    "Описание": "description",
    "Ремонт": "renovation",
    "Площадь комнат, м2": "room_area",
    "Балкон": "balcony",
    "Окна": "windows",
    "Санузел": "bathroom",
    "Можно с детьми/животными": "pets_children_allowed",
    "Дополнительно": "amenities",
    "Название ЖК": "residential_complex",
    "Серия дома": "building_series",
    "Высота потолков, м": "ceiling_height",
    "Лифт": "elevator",
    "Мусоропровод": "garbage_chute",
    "Ссылка на объявление": "listing_url",
    "Валюта": "currency",
    "Цена числом": "price",
    "Общая площадь": "total_area",
    "Цена за м²": "price_per_sqm",
    "rooms": "rooms",
    "room_layout": "room_layout"
})

print(df_rub.columns.tolist())
```

```python
df_rub["metro_station"] = (
    df_rub["metro"]
    .str.extract(r"м\.\s*(.*?)\s*\(")[0]
)

df_rub["metro_walk_minutes"] = pd.to_numeric(
    df_rub["metro"].str.extract(r"\((\d+)\s*мин пешком\)")[0],
    errors="coerce"
)
```

```python
display(df_rub[["metro", "metro_station", "metro_walk_minutes"]].head(10))
```

```python
df_rub["floor"] = pd.to_numeric(
    df_rub["building"].str.extract(r"(\d+)\s*/")[0],
    errors="coerce"
)

df_rub["total_floors"] = pd.to_numeric(
    df_rub["building"].str.extract(r"/\s*(\d+)")[0],
    errors="coerce"
)

display(df_rub[["building", "floor", "total_floors"]].head(5))
```

```python
df_rub["passenger_elevators"] = pd.to_numeric(
    df_rub["elevator"].str.extract(r"Пасс\s*\((\d+)\)")[0],
    errors="coerce"
)

df_rub["freight_elevators"] = pd.to_numeric(
    df_rub["elevator"].str.extract(r"Груз\s*\((\d+)\)")[0],
    errors="coerce"
)

display(df_rub[["elevator", "passenger_elevators", "freight_elevators"]].head(5))
```

```python
df_rub["balconies"] = pd.to_numeric(
    df_rub["balcony"].str.extract(r"Балкон\s*\((\d+)\)")[0],
    errors="coerce"
)

df_rub["loggias"] = pd.to_numeric(
    df_rub["balcony"].str.extract(r"Лоджия\s*\((\d+)\)")[0],
    errors="coerce"
)
display(df_rub[["loggias", "balconies", "balcony"]].sample(10))
```

```python
df_rub["combined_bathrooms"] = pd.to_numeric(
    df_rub["bathroom"].str.extract(r"Совмещенный\s*\((\d+)\)")[0],
    errors="coerce"
)

df_rub["separate_bathrooms"] = pd.to_numeric(
    df_rub["bathroom"].str.extract(r"Раздельный\s*\((\d+)\)")[0],
    errors="coerce"
)
```

```python
display(df_rub[
    [
        "rooms",
        "room_layout",
        "metro_station",
        "metro_walk_minutes",
        "floor",
        "total_floors",
        "passenger_elevators",
        "freight_elevators",
        "balconies",
        "loggias",
        "combined_bathrooms",
        "separate_bathrooms"
    ]
].head())
```

```python
df_rub["children_allowed"] = (
    df_rub["pets_children_allowed"]
    .str.contains("Можно с детьми", na=False)
    .astype(int)
)

df_rub["pets_allowed"] = (
    df_rub["pets_children_allowed"]
    .str.contains("Можно с животными", na=False)
    .astype(int)
)
```

```python
df_rub["furniture"] = (
    df_rub["amenities"]
    .str.contains("Мебель", na=False)
    .astype(int)
)
```

```python
df_rub["washing_machine"] = (
    df_rub["amenities"]
    .str.contains("Стиральная машина", na=False)
    .astype(int)
)
```

```python
df_rub["dishwasher"] = (
    df_rub["amenities"]
    .str.contains("Посудомоечная машина", na=False)
    .astype(int)
)
```

```python
df_rub["air_conditioner"] = (
    df_rub["amenities"]
    .str.contains("Кондиционер", na=False)
    .astype(int)
)
```

```python
display(
    df_rub.sample(5)[[
        "children_allowed",
        "pets_allowed",
        "furniture",
        "washing_machine",
        "air_conditioner",
        "dishwasher"
]])
```

```python
categorical_columns = [
    "room_layout",
    "parking",
    "renovation",
    "windows",
    "garbage_chute",
    "property_type",
    "currency",
]

for column in categorical_columns:
    df_rub[column] = df_rub[column].astype("category").cat.codes
```

```python
numeric_columns = df_rub.select_dtypes(include="number").columns
```

```python
for column in numeric_columns:
    if df_rub[column].isna().any():
        df_rub[column] = df_rub[column].fillna(df_rub[column].median())
```

```python
print("Всего пропусков:", df_rub.isna().sum().sum())
```

```python
df_rub["metro_station"] = df_rub["metro_station"].fillna("unknown")
```
